# Parte 1: Python base

Python es un lenguaje de propósito general muy usado en automatización, análisis de datos, desarrollo web, inteligencia artificial y enseñanza.

## 1. Variables y asignación

Una variable es un nombre que apunta a un valor. En Python no declaramos el tipo de forma explícita; el tipo se infiere a partir del valor asignado.

```python
nombre = "Ana"
edad = 21
altura = 1.68
estudia_python = True

print(nombre)
print(edad)
print(altura)
print(estudia_python)
```

```python
# Podemos consultar el tipo de un valor con type().
print(type(nombre))
print(type(edad))
print(type(altura))
print(type(estudia_python))
```

Los tipos más comunes al empezar son:

- `int`: números enteros.
- `float`: números decimales.

- `bool`: valores lógicos, `True` o `False`.

## 3. Operadores

Python tiene operadores aritméticos, de comparación y lógicos.

```python
a = 10
b = 3

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)
print("División entera:", a // b)
print("Módulo:", a % b)
print("Potencia:", a ** b)
```

## 4. Strings

Los strings representan texto. Son secuencias, por lo que se pueden indexar, recorrer y cortar.

```python
frase = "aprender Python abre muchas puertas"

print(frase.upper())
print(frase.lower())
print(frase.capitalize())
print(frase.replace("Python", "programación"))
print(frase.split())
print(len(frase))
```

## 5. Listas

Una lista es una colección ordenada y mutable. Puede contener valores de cualquier tipo, aunque en análisis de datos suele ser más claro mantener elementos del mismo tipo.

```python
notas = [18, 15, 20, 12, 17]

print(notas)
print(notas[0])
print(notas[-1])
print(notas[1:4])

notas.append(19)
notas.remove(12)

print(notas)
print("Promedio:", sum(notas) / len(notas))
```

## 6. Tuplas

Una tupla es una colección ordenada e inmutable. Se usa cuando queremos agrupar valores que no deberían cambiar.

```python
punto = (4, 7)
x, y = punto

print(punto)
print("x:", x)
print("y:", y)
```

## 7. Diccionarios

Un diccionario guarda pares `clave: valor`. Es una estructura central en Python porque permite representar registros, configuraciones y datos semiestructurados.

```python
estudiante = {
    "nombre": "Luis",
    "edad": 20,
    "materias": ["Programación", "Estadística", "Álgebra"],
    "activo": True,
}

print(estudiante["nombre"])
print(estudiante.get("edad"))

estudiante["promedio"] = 17.8
print(estudiante)
```

```python
for clave, valor in estudiante.items():
    print(clave, "->", valor)
```

## 8. Conjuntos

Un conjunto (`set`) almacena elementos únicos. Es útil para eliminar duplicados y hacer operaciones como unión, intersección y diferencia.

```python
grupo_a = {"Ana", "Luis", "Marta", "Pedro"}
grupo_b = {"Luis", "Pedro", "Sofía", "Elena"}

print("Unión:", grupo_a | grupo_b)
print("Intersección:", grupo_a & grupo_b)
print("Solo A:", grupo_a - grupo_b)
```

## 9. Condicionales

Los condicionales permiten ejecutar código dependiendo de una condición.

```python
nota = 16

if nota >= 18:
    resultado = "excelente"
elif nota >= 10:
    resultado = "aprobado"
else:
    resultado = "reprobado"

print(resultado)
```

## 10. Bucles

Los bucles permiten repetir instrucciones. En Python usamos principalmente `for` para recorrer colecciones y `while` para repetir mientras una condición sea verdadera.

```python
for nota in notas:
    print(nota)
```

```python
contador = 1

while contador <= 5:
    print("Iteración", contador)
    contador += 1
```

## 11. Comprensiones

Las comprensiones permiten crear listas, diccionarios o conjuntos de forma compacta y expresiva.

```python
numeros = [1, 2, 3, 4, 5, 6]
cuadrados = [n ** 2 for n in numeros]
pares = [n for n in numeros if n % 2 == 0]

print(cuadrados)
print(pares)
```

```python
cuadrados_dict = {n: n ** 2 for n in numeros}
print(cuadrados_dict)
```

## 12. Funciones

Una función agrupa instrucciones reutilizables. Ayuda a evitar repetición, mejora la legibilidad y permite probar partes del código por separado.

```python
def calcular_promedio(valores):
    """Calcula el promedio de una lista no vacía de valores numéricos."""
    return sum(valores) / len(valores)

print(calcular_promedio([10, 15, 20]))
```

```python
def clasificar_nota(nota, minimo_aprobatorio=10):
    if nota >= minimo_aprobatorio:
        return "aprobado"
    return "reprobado"

print(clasificar_nota(12))
print(clasificar_nota(8))
print(clasificar_nota(13, minimo_aprobatorio=14))
```

## 13. Alcance de variables

El alcance indica dónde existe una variable. Una variable definida dentro de una función es local a esa función.

```python
valor_global = 100

def mostrar_alcance():
    valor_local = 25
    print("Dentro de la función:", valor_local)
    print("También puedo leer valor_global:", valor_global)

mostrar_alcance()
print("Fuera de la función:", valor_global)
```

## 14. Errores y excepciones

Los errores son parte normal de programar. Python permite capturar excepciones con `try` y `except` para responder de forma controlada.

```python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "No se puede dividir entre cero"

print(dividir(10, 2))
print(dividir(10, 0))
```

## 21. Vectorización y performance

La vectorización consiste en expresar operaciones sobre colecciones completas de datos en lugar de procesar elemento por elemento con bucles de Python.

En Python puro, un bucle ejecuta muchas instrucciones interpretadas. En NumPy y Pandas, muchas operaciones se delegan a código optimizado de bajo nivel. Por eso, una expresión vectorizada suele ser más corta, más legible y más rápida.

Para medir tiempos usaremos `time.perf_counter()`. Los resultados exactos pueden cambiar según la computadora, pero la diferencia de orden de magnitud suele ser clara.

```python
from time import perf_counter
import pandas as pd

N = 1_000_000
lista_valores = list(range(N))
array_valores = np.arange(N)
serie_valores = pd.Series(array_valores)

print("Cantidad de valores:", N)
```

### Caso 1: operación matemática elemento a elemento

Calcularemos la misma expresión para cada valor:

`resultado = x ** 2 + 3 * x + 10`

```python
inicio = perf_counter()
resultado_python = [x ** 2 + 3 * x + 10 for x in lista_valores]
tiempo_python = perf_counter() - inicio

inicio = perf_counter()
resultado_numpy = array_valores ** 2 + 3 * array_valores + 10
tiempo_numpy = perf_counter() - inicio

inicio = perf_counter()
resultado_pandas = serie_valores ** 2 + 3 * serie_valores + 10
tiempo_pandas = perf_counter() - inicio

print(f"Python puro: {tiempo_python:.6f} segundos")
print(f"NumPy vectorizado: {tiempo_numpy:.6f} segundos")
print(f"Pandas vectorizado: {tiempo_pandas:.6f} segundos")
print(f"NumPy fue {tiempo_python / tiempo_numpy:.1f}x más rápido que Python puro")
print(f"Pandas fue {tiempo_python / tiempo_pandas:.1f}x más rápido que Python puro")

print("Resultados equivalentes:", resultado_python[:5] == resultado_numpy[:5].tolist() == resultado_pandas.head().tolist())
```

### Caso 2: filtrado condicional

Ahora compararemos un filtro: conservar solo valores pares mayores que 500000.

```python
inicio = perf_counter()
filtrado_python = [x for x in lista_valores if x % 2 == 0 and x > 500_000]
tiempo_python_filtro = perf_counter() - inicio

inicio = perf_counter()
filtrado_numpy = array_valores[(array_valores % 2 == 0) & (array_valores > 500_000)]
tiempo_numpy_filtro = perf_counter() - inicio

inicio = perf_counter()
filtrado_pandas = serie_valores[(serie_valores % 2 == 0) & (serie_valores > 500_000)]
tiempo_pandas_filtro = perf_counter() - inicio

print(f"Python puro: {tiempo_python_filtro:.6f} segundos")
print(f"NumPy vectorizado: {tiempo_numpy_filtro:.6f} segundos")
print(f"Pandas vectorizado: {tiempo_pandas_filtro:.6f} segundos")
print("Cantidad filtrada en Python:", len(filtrado_python))
print("Cantidad filtrada en NumPy:", len(filtrado_numpy))
print("Cantidad filtrada en Pandas:", len(filtrado_pandas))
```

### Lectura importante

Vectorizar no significa que nunca debamos usar bucles. Los bucles siguen siendo útiles cuando la lógica es pequeña, irregular o difícil de expresar como operación de arreglos.

Pero cuando trabajamos con columnas numéricas, transformaciones repetitivas, filtros y cálculos sobre grandes volúmenes de datos, NumPy y Pandas suelen ser la mejor opción.

# Parte 3: Introducción a Pandas

Pandas es una librería para trabajar con datos tabulares. Sus estructuras principales son:

- `Series`: una columna con índice.
- `DataFrame`: una tabla con filas y columnas.

Pandas es especialmente útil para cargar datos, inspeccionarlos, limpiarlos, transformarlos, agruparlos y prepararlos para análisis.

```python
import pandas as pd

pd.__version__
```

## 28. Series

Una `Series` es parecida a una lista o arreglo, pero con índice y muchas operaciones integradas.

```python
ventas = pd.Series([120, 150, 90, 210], index=["lunes", "martes", "miércoles", "jueves"])

print(ventas)
print(ventas["martes"])
print(ventas.mean())
```

## 29. DataFrames

Un `DataFrame` representa datos en forma de tabla.

```python
datos = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta", "Pedro", "Sofía", "Elena"],
    "edad": [21, 23, 20, 25, 22, 24],
    "ciudad": ["Valencia", "Caracas", "Valencia", "Maracay", "Caracas", "Maracay"],
    "horas_estudio": [12, 8, 15, 5, 10, 9],
    "puntaje": [88, 75, 95, 62, 81, 79],
})

datos
```

## 30. Inspección inicial

Antes de transformar una tabla, conviene inspeccionar su estructura.

```python
datos.head()
```

```python
datos.info()
```

## 31. Selección de columnas y filas

Pandas permite seleccionar columnas por nombre y filas por posición (`iloc`) o por etiqueta/condición (`loc`).

```python
datos["puntaje"]
```

## 32. Crear y modificar columnas

En Pandas es común crear columnas derivadas a partir de columnas existentes.

```python
datos = datos.copy()
datos["puntaje_ajustado"] = datos["puntaje"] + 5
datos["estudia_mucho"] = datos["horas_estudio"] >= 10

datos
```

```python
def clasificar_puntaje(puntaje):
    if puntaje >= 90:
        return "alto"
    if puntaje >= 75:
        return "medio"
    return "bajo"

datos["nivel"] = datos["puntaje"].apply(clasificar_puntaje)
datos
```

## 33. Ordenamiento

Ordenar ayuda a identificar extremos, prioridades o patrones.

```python
datos.sort_values("puntaje", ascending=False)
```

## 34. Valores faltantes

Los valores faltantes aparecen con frecuencia en datos reales. Pandas los representa normalmente como `NaN`.

```python
datos_con_faltantes = pd.DataFrame({
    "producto": ["A", "B", "C", "D", "E"],
    "precio": [10.5, 8.0, None, 12.0, 9.5],
    "unidades": [100, None, 80, 60, None],
    "categoria": ["base", "base", "premium", None, "premium"],
})

datos_con_faltantes
```

```python
datos_con_faltantes.isna().sum()
```

```python
relleno = datos_con_faltantes.copy()
relleno["precio"] = relleno["precio"].fillna(relleno["precio"].median())
relleno["unidades"] = relleno["unidades"].fillna(0)
relleno["categoria"] = relleno["categoria"].fillna("sin_categoria")

relleno
```

## 35. Agrupaciones con groupby

`groupby` permite resumir datos por categorías.

```python
datos.groupby("ciudad")["puntaje"].mean()
```

```python
resumen_ciudad = datos.groupby("ciudad").agg(
    cantidad=("nombre", "count"),
    promedio_puntaje=("puntaje", "mean"),
    max_horas=("horas_estudio", "max"),
)

resumen_ciudad
```

## 36. Tablas dinámicas

Las tablas dinámicas resumen datos cruzando categorías.

```python
pd.pivot_table(
    datos,
    values="puntaje",
    index="ciudad",
    columns="nivel",
    aggfunc="mean",
)
```

## 37. Combinar tablas

En proyectos reales solemos tener información repartida en varias tablas. `merge` permite unirlas usando una clave común.

```python
estudiantes = pd.DataFrame({
    "id_estudiante": [1, 2, 3, 4],
    "nombre": ["Ana", "Luis", "Marta", "Pedro"],
})

resultados = pd.DataFrame({
    "id_estudiante": [1, 2, 3, 4],
    "puntaje_final": [91, 78, 96, 70],
})

estudiantes.merge(resultados, on="id_estudiante", how="inner")
```

## 38. Concatenar tablas

`concat` permite apilar tablas por filas o columnas.

```python
enero = pd.DataFrame({"mes": "enero", "ventas": [100, 120, 90]})
febrero = pd.DataFrame({"mes": "febrero", "ventas": [130, 110, 140]})

pd.concat([enero, febrero], ignore_index=True)
```

## 39. Reemplazos y mapeos

`map` y `replace` son útiles para transformar categorías.

```python
mapa_ciudades = {
    "Valencia": "centro",
    "Caracas": "capital",
    "Maracay": "centro",
}

datos["zona"] = datos["ciudad"].map(mapa_ciudades)
datos
```

## 40. Fechas

Pandas tiene herramientas muy potentes para trabajar con fechas.

```python
asistencias = pd.DataFrame({
    "fecha": ["2026-01-05", "2026-01-12", "2026-02-02", "2026-02-09"],
    "asistentes": [25, 28, 22, 30],
})

asistencias["fecha"] = pd.to_datetime(asistencias["fecha"])
asistencias["mes"] = asistencias["fecha"].dt.month_name()
asistencias["dia_semana"] = asistencias["fecha"].dt.day_name()

asistencias
```

## 42. Pipeline pequeño de limpieza con Pandas

En esta sección hacemos un flujo compacto de limpieza y transformación sobre una tabla de ejemplo.

```python
registros = pd.DataFrame({
    "cliente": ["A", "B", "C", "D", "E", "F"],
    "segmento": ["nuevo", "recurrente", "nuevo", "recurrente", None, "nuevo"],
    "ingreso": [1200, 2500, None, 3100, 1800, 1600],
    "compras": [2, 5, 1, None, 3, 2],
})

registros
```

```python
limpio = (
    registros
    .assign(
        segmento=lambda df: df["segmento"].fillna("sin_segmento"),
        ingreso=lambda df: df["ingreso"].fillna(df["ingreso"].median()),
        compras=lambda df: df["compras"].fillna(0),
    )
    .assign(
        ingreso_por_compra=lambda df: np.where(
            df["compras"] > 0,
            df["ingreso"] / df["compras"],
            0,
        )
    )
    .sort_values("ingreso", ascending=False)
)

limpio
```

## 45. Ejercicios de Pandas

Usa el DataFrame `ventas_tienda` para practicar.

1. Muestra las primeras filas y la información general.
2. Calcula el ingreso como `precio * unidades`.
3. Filtra las ventas con ingreso mayor o igual a 500.
4. Calcula el ingreso total por categoría.
5. Encuentra el producto con mayor ingreso.
6. Rellena los valores faltantes de descuento con 0.
7. Crea una columna `ingreso_neto` restando el descuento.
8. Ordena los datos por ingreso neto de mayor a menor.

```python
ventas_tienda = pd.DataFrame({
    "producto": ["cuaderno", "lápiz", "mochila", "regla", "marcador", "agenda", "carpeta"],
    "categoria": ["papelería", "papelería", "accesorios", "papelería", "papelería", "papelería", "archivo"],
    "precio": [4.5, 1.2, 35.0, 2.0, 3.5, 12.0, 5.0],
    "unidades": [120, 300, 18, 90, 75, 40, 60],
    "descuento": [20, None, 50, 0, None, 15, 10],
})

ventas_tienda
```

```python
# Solución sugerida
ventas_resueltas = ventas_tienda.copy()
ventas_resueltas["descuento"] = ventas_resueltas["descuento"].fillna(0)
ventas_resueltas["ingreso"] = ventas_resueltas["precio"] * ventas_resueltas["unidades"]
ventas_resueltas["ingreso_neto"] = ventas_resueltas["ingreso"] - ventas_resueltas["descuento"]

print(ventas_resueltas.info())
display(ventas_resueltas.head())
display(ventas_resueltas.loc[ventas_resueltas["ingreso"] >= 500])
display(ventas_resueltas.groupby("categoria")["ingreso"].sum().sort_values(ascending=False))
display(ventas_resueltas.sort_values("ingreso_neto", ascending=False))
```
